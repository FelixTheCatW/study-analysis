## Простая идея ООП

Класс — это шаблон.

Объект — это конкретный экземпляр этого шаблона.

Например:

```python
class Student:
    ...
```

`Student` — класс.

```python
student = Student("Анна")
```

`student` — объект.

---

## Прикладное значение

В реальных проектах классы помогают описывать сущности:

- студент;
- товар;
- заказ;
- клиент;
- датасет;
- аналитический отчёт;
- анализатор данных.

Вместо набора разрозненных функций мы можем создать объект, который хранит данные и умеет с ними работать.

---

## Связь с дипломом

В дипломном проекте можно будет использовать классы:

```python
class DatasetAnalyzer:
    ...
```

Такой класс сможет:

- хранить данные;
- считать статистику;
- строить отчёты;
- подготавливать данные для машинного обучения.

## Ячейка 5. TODO: класс Product

### Алгоритм

1. Создайте класс `Product`.
2. Добавьте `name`, `price`, `quantity`.
3. Создайте метод `get_total_price`.
4. Проверьте результат.

````python
class ProductPosition:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def get_total_price(self):
        return self.price * self.quantity


product = ProductPosition("Форель", 1200, 4)
total = product.get_total_price()

print("Общая стоимость:", total)
assert total == 4800
````

## Ячейка 6. TODO: список объектов

````python
products = [
    ProductPosition("Гречка", 200, 4),
    ProductPosition("Индейка", 400, 2),
    ProductPosition("Молоко", 120, 1,)
]

total_sum = 0

for product in products:
    total_sum += product.get_total_price()
    print(product.name, product.get_total_price())
    
print("Итоговая сумма:", total_sum)

assert total_sum == 1720
````

## Ячейка 7. TODO: класс Dataset

2. Добавьте `name` и `values`.
3. Создайте метод `get_size`.
4. Проверьте размер.

````python
class Dataset:
    def __init__(self, name, values):
        self.name = name
        self.values = values

    def get_size(self):
        return len(self.values)


dataset = Dataset("sales", [500, 20, 1.5, 300_000_000, -0, -100500])

print("Название:", dataset.name)
print("Размер:", dataset.get_size())

assert dataset.get_size() == 6
````

## Ячейка 8. TODO: класс DatasetAnalyzer

### Алгоритм

1. Создайте класс `DatasetAnalyzer`.
2. Передайте список чисел.
3. Добавьте методы:
   - get_min;
   - get_max;
   - get_mean.
4. Проверьте вычисления.

````python
import numpy as np

rng = np.random.default_rng(37811)

class DatasetAnalyzer:
    def __init__(self, values):
        self.values = values

    def get_min(self):
        return min(self.values)

    def get_max(self):
        return max(self.values)

    def get_mean(self):
        return sum(self.values) / len(self.values)


analyzer = DatasetAnalyzer(rng.integers(504, 643, 3))

print("Минимум:", analyzer.get_min())
print("Максимум:", analyzer.get_max())
print("Среднее:", analyzer.get_mean())

assert analyzer.get_min() == 621
assert analyzer.get_max() == 621
assert analyzer.get_mean() == 621
````

## Ячейка 9. TODO: метод отчёта

### Алгоритм

1. Создайте класс `DatasetAnalyzerWithReport`.
2. Добавьте методы статистики.
3. Добавьте метод `get_report`.
4. Верните словарь.
5. Проверьте отчёт.

````python
class DatasetAnalyzerWithReport:
    def __init__(self, name, values):
        self.name = name
        self.values = values

    def get_min(self):
        return min(self.values)

    def get_max(self):
        return max(self.values)

    def get_mean(self):
        return sum(self.values) / len(self.values)

    def get_report(self):
        return {
            "dataset": self.name,
            "min": self.get_min(),
            "max": self.get_max(),
            "mean": self.get_mean(),
        }

rng = np.random.default_rng(37811)
analyzer = DatasetAnalyzerWithReport("calories", rng.integers(520, 640, 3))

report = analyzer.get_report()

print(report)

assert report["dataset"] == "calories"
assert report["max"] == 621
````
