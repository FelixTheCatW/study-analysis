---

## Прикладное значение

Без функций код быстро становится длинным и неудобным.

Например, если в проекте много раз нужно считать:

- сумму;
- среднее;
- минимум;
- максимум;
- отчёт,

то лучше один раз написать функцию и использовать её много раз.
