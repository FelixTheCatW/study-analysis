````python
import numpy as np

rng = np.random.default_rng(9)


# TODO: создайте функцию calculate_sum(numbers)
def calculate_sum(numbers):
    # TODO: верните sum(numbers)
    s = sum(numbers)
    return s


# TODO: создайте список values
nums = rng.integers(1, 21, 48)
# TODO: вызовите функцию
result = calculate_sum(nums)
print(f"{' + '.join(map(str, nums))} = {result}")
# TODO: assert result == 10
assert result == 621
````

## Ячейка 7. TODO: функция статистики

### Алгоритм

1. Создайте функцию `calculate_statistics(numbers)`.
2. Верните словарь с ключами:
   - min;
   - max;
   - mean.
3. Проверьте значения через assert.

````python
rng = np.random.default_rng(37093)


# TODO: создайте функцию calculate_statistics(numbers)
def calculate_statistics(numbers):
    # TODO: внутри создайте словарь statistics
    # TODO: верните statistics
    return {
        "min": min(numbers),
        "max": max(numbers),
        "mean": sum(numbers) / len(numbers),
    }

# TODO: вызовите функцию
result = calculate_statistics(rng.integers(600, 650, 3))

# TODO: добавьте assert для min, max, mean
assert result["min"] == 621
assert result["max"] == 621
assert result["mean"] == 621
````
