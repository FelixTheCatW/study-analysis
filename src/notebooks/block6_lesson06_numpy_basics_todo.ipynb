## Что такое NumPy

`NumPy` — библиотека Python для работы с числовыми массивами.

Обычный список Python:

```python
values = [10, 20, 30]
```

NumPy-массив:

```python
array = np.array([10, 20, 30])
```

NumPy позволяет выполнять операции сразу со всем массивом:

```python
array * 2
```

## Ячейка 9. TODO: NumPy в функции анализа

### Алгоритм

1. Создайте функцию `analyze_array`.
2. Внутри преобразуйте данные в NumPy-массив.
3. Верните словарь со статистикой.
4. Проверьте значения.

````python
# TODO: создайте функцию analyze_array(values)
def analyze_array(values):
    array = np.array(values)
# TODO: внутри создайте array = np.array(values)
    return {
        "min": np.min(array),
        "max": np.max(array),
        "mean": np.mean(array),
        "sum": np.sum(array)
    }
# TODO: верните словарь min, max, mean, sum
vals = np.random.default_rng(5).integers(10, 100, 10)
print(vals)
result = analyze_array(values=vals)
# TODO: вызовите функцию
print(result)
# TODO: assert
````
