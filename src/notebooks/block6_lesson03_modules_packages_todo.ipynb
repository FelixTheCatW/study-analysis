### Что изучаем

На прошлых занятиях мы создавали функции.

Теперь научимся организовывать код как настоящий проект:

- создавать модули;

- разделять код по файлам;
- строить структуру проекта;
- готовить код к VS Code и дипломному проекту.

### Почему это важно

В реальном проекте нельзя хранить сотни строк в одном файле.

Обычно проект выглядит так:

```text
project/
│
├── src/
│   ├── file_utils.py
│   ├── statistics_utils.py
│   ├── report_utils.py
│   └── main.py
│
└── data/
```

### Связь с дипломом

В дальнейшем проект DataAnalyzer будет состоять из модулей:

```python
load_dataset()
calculate_statistics()
generate_report()
```

Каждая функция будет жить в отдельном файле.

### Алгоритм

````python
def analyze(values):
    return {"mean": sum(values) / len(values), "max": max(values)}


print(analyze([10, 20, 30]))
````
