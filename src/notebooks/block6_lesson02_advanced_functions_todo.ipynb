## Ячейка 3. TODO: несколько параметров

````python
import numpy as np

rng = np.random.default_rng(338)
# TODO: создайте функцию calculate_mean(a,b,c)
def calculate_mean(a, b, c):
    return (a + b + c) / 3


result = calculate_mean(*rng.integers(1, 1000, 3))

assert result == 621
````

## Ячейка 4. TODO: *args

````python
rng = np.random.default_rng(24)
# TODO: создайте функцию calculate_sum(*args)
def calculate_mean(*args):
    return sum(args)


result = calculate_mean(*rng.integers(50, 150, 6))

assert result == 621
````

## Ячейка 5. TODO: **kwargs

````python
def print_info(**kwargs):
    return kwargs


def print_args(city, name):
    print(f"city: {city}, name: {name}")
    return {"name": name, "city": city}


info = print_info(name="Семён", city="Орёл")

print(info)

out = print_args(**info)
assert out == info
print_args(*info)

(n, nv), (c, cv) = info.items()


assert info[n] == nv
````

## Ячейка 10. TODO: функция дипломного проекта

````python
rng = np.random.default_rng(38)


# TODO: создайте analyze_dataset(name, *values)
def analyze_dataset(name, *values):
    stats = statistics(*values)
    report = {"dataset": name, **stats}    
    report |= statistics(*values)
    return report

report = analyze_dataset("sales", *rng.integers(610, 630, 10))

print(report)

assert report["dataset"] == "sales"
````
