### Что изучаем

На прошлом занятии мы создали простые функции.

Сегодня изучаем:

- параметры по умолчанию;
- именованные параметры;
- `*args`;
- `**kwargs`;
- универсальные функции;
- проектирование функций для анализа данных.

### Практическое значение

В реальных проектах анализа данных функции должны работать с разными наборами данных.

Одна хорошая функция может заменить десятки похожих функций.
