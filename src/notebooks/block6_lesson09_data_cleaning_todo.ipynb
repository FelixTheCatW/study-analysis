# Блок 6. Погружение в Python и анализ данных
## Занятие 9. Очистка данных в Pandas

## Цель занятия

После загрузки данных необходимо убедиться, что они пригодны для анализа.

На практике данные часто содержат:

- пропущенные значения;
- дубликаты;
- ошибки ввода;
- пустые строки;
- неверные типы данных.

Сегодня изучаем:

- поиск пропусков;
- удаление пропусков;
- заполнение пропусков;
- поиск дубликатов;
- удаление дубликатов;
- подготовку данных к машинному обучению.

---

## Почему очистка данных важна

Даже самая хорошая модель машинного обучения будет давать плохие результаты на грязных данных.

Поэтому правило аналитика:

```text
Сначала очистка данных.
Потом анализ.
Потом машинное обучение.
```
