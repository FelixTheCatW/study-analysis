## Ячейка 1. Импорт библиотек

### Алгоритм

Подключаем Pandas и загружаем датасет.

````python
# TODO: выполните задание ячейки 1
import pandas as pd

path = "d:\\study\\excel\\block05_exam_project\\data\\diaries_dirty.parquet"
df = pd.read_parquet(path)

print(df.shape)

assert len(df) > 0
````

## Ячейка 2. Изучение структуры данных

### Алгоритм

Смотрим размер таблицы, колонки и типы данных.

````python
print(df.shape)
print(df.columns.tolist())

df.info()

assert df.shape[0] > 1000
````

## Ячейка 3. Поиск пропусков

### Алгоритм

Используем isna() и sum().

````python
missing = df.isna().sum()

print(missing)

assert "Calories" in missing.index
````

## Ячейка 4. Просмотр строк с пропусками

### Алгоритм

Находим записи с пустыми значениями.

````python
rows_with_nan = df[df.isna().any(axis=1)]

display(rows_with_nan.head())
print(list(rows_with_nan.columns))
assert len(rows_with_nan) > 0
````

## Ячейка 5. Удаление пропусков

### Алгоритм

Создаем новую очищенную таблицу.

````python
df_dropna = df.dropna()

print(df.shape)
print(df_dropna.shape)

assert len(df_dropna) < len(df)
````

## Ячейка 6. Заполнение пропусков

### Алгоритм

Заполняем пропуски средним значением.

````python
df_fillna = df.copy()

cols_with_nan = df_fillna.columns[df_fillna.isna().any()].tolist()
print("Колонки с NaN:", cols_with_nan)

for col in cols_with_nan:
    df_fillna[col] = df_fillna[col].fillna(0)

print(df_fillna.isna().sum().sum())
assert df_fillna.isna().sum().sum() == 0
````

## Ячейка 7. Создание искусственных дубликатов

### Алгоритм

Добавляем несколько повторяющихся строк.

````python
duplicates_df = pd.concat([df.head(5), df.head(5)])

print(duplicates_df.shape)

assert duplicates_df.shape[0] == 10
````

## Ячейка 8. Поиск и удаление дубликатов

### Алгоритм

Используем duplicated() и drop_duplicates().

````python
duplicate_count = duplicates_df.duplicated().sum()

clean_df = duplicates_df.drop_duplicates()

print("Дубликатов:", duplicate_count)
print("После очистки:", clean_df.shape)

assert duplicate_count == 5
````

## Ячейка 9. Подготовка данных к анализу

### Алгоритм

Выбираем важные колонки.

````python
prepared_df = df_fillna[["user_id", "date", "Calories", "Carbs", "Fat", "Protein"]]

display(prepared_df.head())
print(prepared_df.shape)
assert prepared_df.shape[1] == 6 
````

## Ячейка 10. Итоговый отчет по очистке

### Алгоритм

Формируем отчет по качеству данных.

````python
report = {
    "rows_before": len(df),
    "rows_after_dropna": len(df_dropna),
    "missing_before": int(df.isna().sum().sum()),
    "missing_after_fillna": int(df_fillna.isna().sum().sum()),
    "duplicates_removed": int(duplicate_count),
}

for k, v in report.items():
    print(k, ":", v)

assert report["missing_after_fillna"] == 0
````
