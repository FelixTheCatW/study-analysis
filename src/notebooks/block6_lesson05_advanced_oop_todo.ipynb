## Ячейка 1. Класс Dataset

### Алгоритм

Создаем класс для хранения данных.

```python
import numpy as np

rng = np.random.default_rng()

# TODO: выполните задание ячейки 1
class Dataset:
    def __init__(self, name, values):
        self.name = name
        self.values = values


dataset = Dataset("calories", rng.integers(10, 700, size=100))

assert dataset.name == "calories"
```

## Ячейка 2. Класс Analyzer

### Алгоритм

Создаем анализатор данных.

```python
# TODO: выполните задание ячейки 2
class Analyzer:
    def __init__(self, values):
        self.values = values

    def mean(self):
        return sum(self.values) / len(self.values)


a = Analyzer(rng.integers(10, 500, 100))
assert a.mean() == 20
```
