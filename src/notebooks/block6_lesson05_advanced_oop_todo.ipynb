## Ячейка 1. Класс Dataset

### Алгоритм

Создаем класс для хранения данных.

```python
import numpy as np

rng = np.random.default_rng(20)

# TODO: выполните задание ячейки 1
class Dataset:
    def __init__(self, name, values):
        self.name = name
        self.values = values


dataset = Dataset("calories", rng.integers(10, 700, size=100))

assert dataset.name == "calories"
```

## Ячейка 2. Класс Analyzer

### Алгоритм

Создаем анализатор данных.

```python
# TODO: выполните задание ячейки 2
rng = np.random.default_rng(20)
class Analyzer:
    def __init__(self, values):
        self.values = values

    def mean(self):
        return sum(self.values) / len(self.values)

a = Analyzer(rng.integers(10, 500, 100))
print(a.mean())
assert a.mean() - 255.41 < 1
```

## Ячейка 3. Класс Report

### Алгоритм

Создаем объект отчета.

```python
# TODO: выполните задание ячейки 3
class Report:
    def __init__(self, title):
        self.title = title

report = Report("Calories Report")

assert report.title == "Calories Report"
```

## Ячейка 4. Композиция классов

### Алгоритм

Analyzer получает Dataset.

```python
rng = np.random.default_rng(21)
dataset = Dataset("Calories", rng.integers(10,200,300))
analyzer = Analyzer(dataset.values)

assert  analyzer.mean() - 106 < 1
```

## Ячейка 5. Методы минимума и максимума

### Алгоритм

Добавляем новые методы.

```python
# TODO: выполните задание ячейки 5
class Analyzer:
    def __init__(self, values):
        self.values = values

    def minimum(self):
        return min(self.values)
    def maximum(self):
        return max(self.values)
rng = np.random.default_rng(22)
a = Analyzer(rng.integers(10, 100, 99))

print(a.minimum(), a.maximum())

assert a.minimum() == 10
assert a.maximum() == 98
```

## Ячейка 6. Генерация отчета

### Алгоритм

Возвращаем словарь отчета.

```python
# TODO: выполните задание ячейки 6
class Analyzer:
    def __init__(self, values):
        self.values = values

    def report(self):
        return {
            "min": min(self.values),
            "max": max(self.values)
        }

r = Analyzer(rng.integers(10, 20, 30)).report()

assert r["max"] == 19
```

## Ячейка 7. DatasetAnalyzer

### Алгоритм

Объединяем Dataset и Analyzer.

```python
class DatasetAnanlyzer:
    def __init__(self, dataset):
        self.dataset = dataset

    def report(self):
        return {
            "dataset": self.dataset.name,
            "count": len(self.dataset.values)
        }

da = DatasetAnanlyzer(dataset)

assert da.report()["dataset"] == "Calories"
```

## Ячейка 8. Работа с продажами

### Алгоритм

Используем DataAnalyzer.

```python
calories = Dataset("calories", rng.integers(100, 300, 50))

da = DatasetAnanlyzer(calories)
print(da.report())

assert da.report()["count"] == 50
```

## Ячейка 9. Работа с тренировками

### Алгоритм

Переиспользуем классы.

```python
training = Dataset("training", rng.integers(50, 130, 100))

da = DatasetAnanlyzer(training)

assert  da.report()["count"] == 100
```
