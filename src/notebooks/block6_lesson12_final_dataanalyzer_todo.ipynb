## Ячейка 9. TODO: класс DataAnalyzer

### Алгоритм

1. Создайте класс `DataAnalyzer`.
2. Добавьте `__init__`.
3. Добавьте метод `get_numeric_statistics`.
4. Добавьте метод `get_group_report`.
5. Создайте объект и проверьте методы.

````python
class DataAnalyzer:
    def __init__(self, dataframe):
        self.dataframe = dataframe

    def get_numeric_statistics(self, column):
        values = self.dataframe[column].to_numpy()
        return {
            "min": np.min(values),
            "max": np.max(values),
            "mean": np.mean(values)
        }

    def get_group_report(self, group_column, value_column):
        report = (
            self.dataframe.groupby(group_column)[value_column]
            .agg(["count", "mean", "min", "max"])
            .sort_values(by="mean", ascending=False)
        )
        return report


analyzer = DataAnalyzer(df_clean)

class_stats = analyzer.get_numeric_statistics("median_house_value")
class_group_report = analyzer.get_group_report("ocean_proximity", "median_house_value")

print(class_stats)
display(class_group_report)

assert class_stats["max"] >= class_stats["min"]
assert class_group_report.shape[0] > 0
````
