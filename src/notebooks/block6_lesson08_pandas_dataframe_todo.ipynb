## Ячейка 9. TODO: статистика и группировка

### Алгоритм

1. Используйте `describe()`.
2. Сгруппируйте данные по `ocean_proximity`.
3. Найдите среднюю стоимость жилья.
4. Отсортируйте результат.

````python
# TODO: создайте stats
stats = df.describe()
# TODO: покажите stats
display(stats)
# TODO: создайте group_report
group_report = (
    df.groupby('user_id')["total_calories"]
    .mean()
    .sort_values(ascending=False)
)
# TODO: выведите group_report
print("Средняя потребление по пользователям")
# TODO: assert
display(group_report.head())
assert "mean" in stats.index
assert len(group_report) > 0
````

## Ячейка 10. TODO: мини-проект Pandas

### Алгоритм

1. Создайте словарь `report`.
2. Постройте bar chart по группам.
3. Создайте список из 5 выводов.
4. Проверьте результат.

````python
import matplotlib.pyplot as plt

report = {
    "rows": len(df),
    "columns": df.shape[1],
    "mean_house_value": df["total_calories"].mean(),
    "max_house_value": df["total_calories"].max(),
    "mean_income": df["total_fat"].mean(),
}

for key, value in report.items():
    print(key, ":", value)

# avg_calories = df.groupby("user_id")["total_calories"].mean().nlargest(5)

fig, ax = plt.subplots(figsize=(9, 5))
group_report.nlargest(5).plot(kind="barh", ax=ax, color="steelblue")
ax.set_title("Топ-5 пользователей по среднему потреблению калорий")
ax.set_xlabel("Средние калории")
ax.set_ylabel("user_id")
ax.invert_yaxis()  # самый большой сверху
plt.tight_layout()
plt.show()

conclusions = [
    "Pandas позволяет удобно загружать и анализировать таблицы.",
    "DataFrame похож на таблицу Excel.",
    "Фильтрация помогает выбирать нужные строки.",
    "Группировка помогает строить отчёты.",
    "Pandas является основой подготовки данных для ML.",
]

for item in conclusions:
    print("-", item)

assert report["rows"] > 1000
assert len(conclusions) == 5
````
